## Class balance

```python
dv.class_table(DS).pivot_table(index="class", columns="scheme", values="rows").fillna(0).astype(int)
```

```python
dv.class_bars(DS, "all23");
```

Four orders of magnitude between `normal` and `spy`. This is why every
supervised head uses class-balanced loss and why macro-F1, not accuracy, is the
headline: accuracy is dominated by `normal` and `neptune`.

## How much each mode count keeps

```python
dv.pca_variance(DS.X, marks=(12, 20, 24, 32, 64));
```

The reservoir only ever sees these components. If a mode count already keeps
most of the variance, the problem is close to linear in them and a reservoir has
little room to add anything (see the PCA rule in `40_images_and_forecasting`).
