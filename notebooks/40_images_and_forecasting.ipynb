## 2. The PCA rule

Gain correlates strongly and negatively with the variance PCA(12) retains: when
twelve components already explain the data, lifting it into 220 dimensions only
adds noise. A seconds-long screen before trying a new dataset:
`PCA(n_modes).fit(X).explained_variance_ratio_.sum()` -- below ~0.65 worth trying,
above ~0.80 the reservoir will probably lose.

The reservoir is memoryless (measurement collapses the state, photons leave), so
`p_t = f(x_t)`. Delay embedding and a feedback loop add memory from outside; the
classical ESN still wins. Position forecasting as a limits study.
